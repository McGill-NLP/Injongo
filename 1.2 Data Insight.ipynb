### Entity Analysis

```python
from mlds.analysis_data import version_compare
import plotly.io as pio

pio.renderers.default = "jupyterlab+png"


combined_unreviewed = []
combined_reviewed = []
for lan in sorted(LANGUAGES):
    unreviewed = entity_manager.load_data(lan, reviewed=False)
    reviewed = entity_manager.load_data(lan, reviewed=True)

    combined_unreviewed.extend(unreviewed)
    combined_reviewed.extend(reviewed)

result = version_compare(
    combined_unreviewed, combined_reviewed, prefix="All Languages: "
)
result = version_compare(
    combined_unreviewed, combined_reviewed, majority=True, prefix="All Languages: "
)
```
