```python
from mlds import data_loader
import pandas as pd
from pprint import pprint

entity_manager = data_loader.EntityDataManager(data_folder="data/json")
intent_manager = data_loader.IntentDataManager(
    data_folder="data/UtteranceGen/submission"
)
slot_manager = data_loader.SlotDataManager(data_folder="data/output")
```

### Entity Analysis

```python
import plotly.io as pio

pio.renderers.default = "jupyterlab+png"

from mlds.analysis_data import version_compare
from mlds.data_loader import LANGUAGES


combined_unreviewed = []
combined_reviewed = []
for lan in sorted(LANGUAGES):
    unreviewed = entity_manager.load_data(lan, reviewed=False, merged=True)
    reviewed = entity_manager.load_data(lan, reviewed=True, merged=True)

    combined_unreviewed.extend(unreviewed)
    combined_reviewed.extend(reviewed)

result = version_compare(
    combined_unreviewed, combined_reviewed, prefix="[" + "all" + "]\t", merged=True
)
result = version_compare(
    combined_unreviewed, combined_reviewed, majority=True, prefix="[" + "all" + "]\t", merged=True
)
```

### Joint Analysis

```python
from mlds.data_loader import INTENTS


def sanity_check(data: pd.DataFrame, n: int = 80):
    # total have 3200 samples
    # 40 intent * 80 samples
    print(len(data))
    print(len(data["intent"].unique()))
    print(all(data.groupby("intent").size() == n))
    print(set(data["intent"].unique()) - set(INTENTS))


for lan in sorted(LANGUAGES):
    print(lan)

    merged = slot_manager.load_data(lan, split="full")
    sanity_check(merged)

    merged = slot_manager.load_data(lan, split="split")
    for split in ["train", "test", "dev"]:
        print(split, end="\t")
        sanity_check(merged[split], {"train": 56, "test": 16, "dev": 8}[split])
```
