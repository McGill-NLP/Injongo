```python
import pandas as pd
from mlds import data_loader
import warnings, re, os
warnings.filterwarnings('ignore')
from transformers import pipeline
from typing import List


translater = pipeline("translation", model="facebook/nllb-200-3.3B", device="cuda:0")
manager = data_loader.SlotDataManager(data_folder="data/output")
mapper = data_loader.NLLB_MAPPER
```

```python
def translate(source_language: str, target_language: str, texts: List[str]) -> List[str]:
    """Translate texts from source language to target language."""
    print(f"Translating {len(texts)} texts from {source_language} to {target_language}")
    sample = texts[:1]
    print(translater(sample, src_lang=source_language, tgt_lang=target_language))
    return [content['translation_text']  for content in translater(texts, src_lang=source_language, tgt_lang=target_language, batch_size=128)]
    # return [translater(text, src_lang=source_language, tgt_lang=target_language)[0] for text in texts]

for language in data_loader.LANGUAGES:
    data = manager.load_data(language, "full")
    # domain,intent,text,spans,logical_form
    # banking,balance,ብዙ ብር ያለው የትኛው አካውንቴ ነው,3:5:SL:CURRENCY,[IN:balance [SL:CURRENCY ብር] ]
    # banking,balance,የባንክ አካውንቶቼ ውስጥ አሁን ያለኝን ቀሪ ሂሳብ ንገሩኝ።,,[IN:balance ]
    # banking,balance,በባንክ አካውንቴ አሁን ያለኝ አጠቃላይ ገንዘብ ምን ያህል ነው።,,[IN:balance ]
    
    data['text'] = translate(mapper[language], mapper["eng"], data['text'].tolist())
    # break
    data.to_csv(f"data/output/{language}_eng.csv", index=False)
data
```
